```python
!pip install -q apify-client pandas

from getpass import getpass
import pandas as pd
from apify_client import ApifyClient

APIFY_TOKEN = getpass("Enter your Apify API Token (from https://apify.com/unitbytes/goofish-xianyu-search-scraper?fpr=939u3w&fp_sid=gh_goofish_api_colab): ").strip()
if not APIFY_TOKEN:
    raise ValueError("Please enter a valid Apify API token to run live extraction.")

client = ApifyClient(APIFY_TOKEN)

run_input = {
    "keyword": "iPhone 15",
    "maxItems": 10,
    "detailLevel": "full",
    "sortBy": "newest"
}

print("Running unitbytes/goofish-xianyu-search-scraper...")
run = client.actor("unitbytes/goofish-xianyu-search-scraper").call(run_input=run_input)
items = list(client.dataset(run["defaultDatasetId"]).iterate_items())

df_live = pd.DataFrame(items)
print(f"Extracted {len(df_live)} live records!")
df_live.to_csv("goofish-xianyu-search-scraper-results.csv", index=False)
df_live.head(10)
```

### ⚡ Step 3: Call the Synchronous `<500ms` Goofish (闲鱼) Standby REST API
Test the real-time synchronous `GET /v1/search` endpoint on `https://unitbytes--goofish-scraper.apify.actor`:


```python
from getpass import getpass
import requests
import pandas as pd

APIFY_TOKEN = getpass("Enter your Apify API Token: ").strip()
resp = requests.get(
    "https://unitbytes--goofish-scraper.apify.actor/v1/search",
    headers={"Authorization": f"Bearer {APIFY_TOKEN}"},
    params={"keyword": "ccd相机", "page": 1, "sort_by": "default"},
    timeout=30,
)
resp.raise_for_status()
pd.DataFrame(resp.json().get("data", {}).get("items", [])).head()
```
